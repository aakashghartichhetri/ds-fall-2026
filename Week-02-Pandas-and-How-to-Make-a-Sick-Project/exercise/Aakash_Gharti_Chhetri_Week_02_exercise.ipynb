But what about the weird non-numeric values in the column obviously meant for numeric data?

Notice the data type of that column from when you got the types of all the columns?

### Clean the Data

Well, we can’t get our answers with what we currently have, so you’ll have to make some changes.

Let’s not worry about anything below the first 5 values for now. Convert the top 5 to a single canonical value―say, `'US'`, which is nice and short.

```python
# Get the five most common country spellings.
top_5_countries = (
    df_salary_tech["country"]
    .value_counts()
    .head(5)
    .index
)

# Change those common U.S. spellings to "US".
df_salary_tech.loc[
    df_salary_tech["country"].isin(top_5_countries),
    "country"
] = "US"

```

Have a look at the count of each unique country again now.

```python
# Count again.
df_salary_tech["country"].value_counts().head(15)
```

Did you notice anything interesting?

```python
# 🚀 Extra credit: Resolve [most of] those anomalous cases too without exhaustively taking every variant literally into account.
us_pattern = (
    r"(?i)^\s*"
    r"(u\.?s\.?a?|united states(?: of america)?|america)"
    r"\s*$"
)

mask_us = (
    df_salary_tech["country"]
    .astype(str)
    .str.strip()
    .str.match(us_pattern, na=False)
)

# Standardize matching values to "US".
df_salary_tech.loc[mask_us, "country"] = "US"
```

```python

# 🚀 Extra credit: If you’ve resolved it, let’s see how well you did by counting the number of instances of each unique value.
df_salary_tech["country"].value_counts().head(20)
```

It’s looking good so far. Let’s find out the minimum, mean, and maximum (in that order) salary by state, sorted by the mean in descending order.

```python
# Clean salary and calculate min, mean, and max by U.S. state.
df_salary_tech["salary"] = pd.to_numeric(
    df_salary_tech["salary"].astype(str).str.replace(r"[$,]", "", regex=True),
    errors="coerce"
)

salary_by_state = (
    df_salary_tech[df_salary_tech["country"] == "US"]
    .groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)

salary_by_state
```

```python
# Convert timestamp to a date/time type.
df_salary_tech["timestamp"] = pd.to_datetime(
    df_salary_tech["timestamp"],
    errors="coerce"
)

# Keep only responses from 2021 and 2022.
df_salary_tech_2021_2022 = df_salary_tech[
    df_salary_tech["timestamp"].dt.year.isin(
        [2021, 2022]
    )
].copy()

# Calculate salary statistics by state.
salary_by_state_2021_2022 = (
    df_salary_tech_2021_2022[
        df_salary_tech_2021_2022["country"] == "US"
    ]
    .groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)

salary_by_state_2021_2022
```

## Bonus

Clearly, we do not have enough data to produce useful figures for the level of specificity you’ve now reached. What do you notice about Delaware and West Virginia?

Let’s back out a bit and return to `df_salary` (which was the loaded data with renamed columns but *sans* filtering).

### Bonus #0

Apply the same steps as before to `df_salary`, but do not filter for any specific industry. Do perform the other data cleaning stuff, and get to a point where you can generate the minimum, mean, and maximum by state.

```python
# Keep USD responses from all industries.
df_salary_all = df_salary[
    df_salary["currency"] == "USD"
].copy()

# Find common U.S. country spellings.
us_pattern = (
    r"(?i)^\s*"
    r"(u\.?s\.?a?|united states(?: of america)?|america)"
    r"\s*$"
)

mask_us = (
    df_salary_all["country"]
    .astype(str)
    .str.strip()
    .str.match(us_pattern, na=False)
)

# Standardize the country name.
df_salary_all.loc[
    mask_us,
    "country"
] = "US"


# Convert salary text into numeric values.
df_salary_all["salary"] = pd.to_numeric(
    df_salary_all["salary"]
    .astype(str)
    .str.replace(r"[$,]", "", regex=True),
    errors="coerce"
)


# Calculate salary statistics across all industries.
all_salary_by_state = (
    df_salary_all[
        df_salary_all["country"] == "US"
    ]
    .groupby("state")["salary"]
    .agg(["min", "mean", "max"])
    .sort_values("mean", ascending=False)
)

all_salary_by_state
```

### Bonus #2

Filter out the non-single-states (e.g., `'California, Colorado'`) in the most elegant way possible (i.e., *not* by blacklisting all the bad values).

```python
# List the 50 valid U.S. state names.
valid_states = {
    "Alabama", "Alaska", "Arizona", "Arkansas",
    "California", "Colorado", "Connecticut",
    "Delaware", "Florida", "Georgia", "Hawaii",
    "Idaho", "Illinois", "Indiana", "Iowa",
    "Kansas", "Kentucky", "Louisiana", "Maine",
    "Maryland", "Massachusetts", "Michigan",
    "Minnesota", "Mississippi", "Missouri",
    "Montana", "Nebraska", "Nevada",
    "New Hampshire", "New Jersey", "New Mexico",
    "New York", "North Carolina", "North Dakota",
    "Ohio", "Oklahoma", "Oregon", "Pennsylvania",
    "Rhode Island", "South Carolina",
    "South Dakota", "Tennessee", "Texas",
    "Utah", "Vermont", "Virginia", "Washington",
    "West Virginia", "Wisconsin", "Wyoming"
}

# Keep only responses with one valid state.
df_salary_single_state = df_salary_all[
    (df_salary_all["country"] == "US")
    & (df_salary_all["state"].isin(valid_states))
].copy()
```

### Bonus #3

Show the quantiles instead of just minimum, mean, and maximum―say 0%, 5%, 25%, 50%, 75%, 95%, and 100%. Outliers may be deceiving.

Sort by whatever interests you―like say the *50th* percentile.

And throw in a count by state too. It would be interesting to know how many data points contribute to the figures for each state. (*Hint: your nice formatting from Bonus #1 might not work this time around.* 😜)

```python
# Percentiles show how salaries are distributed in each state.
quantiles = [
    0,
    0.05,
    0.25,
    0.50,
    0.75,
    0.95,
    1.00
]

salary_quantiles = (
    df_salary_single_state
    .groupby("state")["salary"]
    .quantile(quantiles)
    .unstack()
)

# Give the percentile columns easier names.
salary_quantiles.columns = [
    "0%",
    "5%",
    "25%",
    "50%",
    "75%",
    "95%",
    "100%"
]

# Add the number of responses for each state.
salary_quantiles["count"] = (
    df_salary_single_state
    .groupby("state")["salary"]
    .count()
)

# Sort states by median salary.
salary_quantiles.sort_values(
    "50%",
    ascending=False
)
```
